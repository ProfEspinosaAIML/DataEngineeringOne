Building on the concepts of actions, insights, and predictions in AI and Data Science, we now explore ML pipelines and Data Engineering Lifecycles. 

These frameworks ensure efficient data handling, from collection to actionable outcomes, streamlining the process of transforming raw data into valuable predictions and insights.

Pipelines and lifecycles are structured sequences of processes designed to manage data flow and model development. 

ML pipelines focus on automating machine learning tasks, while Data Engineering Lifecycles emphasize data infrastructure management.

![ML Pipeline](images/ML_Pipeline.png)

The Data Engineering Life Cycle consists of 5 stages:

| Stage | Key points |
| --- | --- |
| Generation | A source system is the origin of the data used in the DE lifecycle. Engineers consume data from source systems and must keep it current while handling multiple source systems. |
| Storage | Choosing the right storage solution is essential. This is often the most complex stage because of cloud storage options, query capabilities, and the need to support hot, warm, and cold data access patterns. |
| Ingestion | Ingestion is the process of importing data from multiple sources. It is a major bottleneck in the lifecycle and often depends on whether data arrives in batch or streaming mode. |
| Transformation | This is the stage where data changes from its original form into something useful for downstream use cases. It is where raw data starts to create real value. |
| Serving | This is the most exciting stage, where data is used to generate insights through analytic tools, machine learning, and reverse ETL. |

![DE Life Cycle](images/DE_Life_Cycle.png)

## Part 2 - Perform some Exploratory Data Analysis on the basic Data Set



#### Query 1
Add Python code to run a query on the existing "movies" DataFrame for extracting movies released in 2012

````python
# Make sure the date column is in datetime format
movies["release_date"] = pd.to_datetime(movies["release_date"], errors="coerce")

# Keep only the movies released in 2012
movies_2012 = movies[movies["release_date"].dt.year == 2012].copy()

# Show the results
movies_2012.head()
````

#### Query 2
Add Python code to run a query on the existing "movies" DataFrame for extracting movies with revenue over 1M

````python
# Filter for movies with revenue over 1 million
movies_revenue_over_1m = movies[movies["revenue"] > 1000000].copy()

# Show the results
movies_revenue_over_1m.head()
````

#### Query 3
Add Python code to run a query on the existing "movies" DataFrame for extracting movies with popularity over 100

NOTE: **What does popularity over 100 mean??**

````python
# Filter for movies with popularity greater than 100
movies_popularity_over_100 = movies[movies["popularity"] > 100].copy()

# Show the results
movies_popularity_over_100.head()
````

#### Query 4

Write Python code to perform statistical analysis to find out of there is a correlation between revenue and popularity. 
Do the necessary queries and plot a chart using MatplotLib where revenue is the independent variable in the X axis and popularity is the dependent varianle in the Y axis.

````python
# Create a smaller dataset with the needed columns
movies_corr = movies[["revenue", "popularity"]].dropna().copy()

# Check correlation between revenue and popularity
correlation = movies_corr["revenue"].corr(movies_corr["popularity"])
print(f"Correlation between revenue and popularity: {correlation:.4f}")

# Plot revenue (X-axis) against popularity (Y-axis)
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(movies_corr["revenue"], movies_corr["popularity"], alpha=0.6)
plt.title("Revenue vs Popularity")
plt.xlabel("Revenue")
plt.ylabel("Popularity")
plt.grid(True)
plt.show()
````
