### Why Use a Virtual Environment?

A Python virtual environment (`venv`) creates an isolated workspace for a project, allowing the project to use its own Python packages and versions without affecting other projects or the system-wide Python installation. This improves reproducibility, avoids dependency conflicts, and makes it easier for others to run the project using the same package versions. In data science and machine learning projects, virtual environments are especially important because different notebooks may require different library versions (e.g., NumPy, Pandas, Scikit-learn, TensorFlow).

### Typical Workflow

1. Open the existing project folder in Visual Studio Code.
2. Create a virtual environment inside the project folder.
3. Activate the virtual environment.
4. Install required packages into the environment.
5. Select the environment as the Python interpreter in VS Code.
6. Select the same environment as the Jupyter kernel for notebook execution.
7. Save dependencies to a `requirements.txt` file when needed.
8. Reactivate the environment whenever returning to the project.

### Basic Setup Instructions

Create the virtual environment from the VS Code terminal:

```bash
python -m venv .venv
```

Activate it:

**Windows (PowerShell)**

```powershell
.\.venv\Scripts\Activate.ps1
```

**Windows (Command Prompt)**

```cmd
.venv\Scripts\activate
```

**macOS/Linux**

```bash
source .venv/bin/activate
```

Install packages:

```bash
pip install numpy pandas matplotlib jupyter
```

Select the interpreter in VS Code:

- Press **Ctrl+Shift+P**.
- Choose **Python: Select Interpreter**.
- Select the interpreter inside the `.venv` folder.

For Jupyter notebooks:

- Click **Select Kernel** in the notebook toolbar.
- Choose the kernel associated with the `.venv` environment.

Optionally save dependencies:

```bash
pip freeze > requirements.txt
```

Later, recreate the environment on another machine using:

```bash
pip install -r requirements.txt
```

A common project structure is:

```text
my-project/
│
├── .venv/
├── notebooks/
├── data/
├── src/
├── requirements.txt
└── README.md
```

Keeping the virtual environment inside the project folder makes the project self-contained, easier to manage, and more reproducible across different computers and development environments.

````python
key = 2
match key:
    case 1:
        print("key is equal to 1")
    case 2:
        print("key is equal to 2")
    case _:
        print("default case")
````

## Data Structures

### Lists

**Definition**: A list is an ordered, mutable collection of elements.

**Characteristics:**
* Preserves the order of insertion.
* Allows duplicate values.
* Elements can be changed (MUTABLE).
* Indexed (you can access items using positions).
* Use Case: When you need an ordered collection with possible duplicates and the ability to modify items.

````python
courses = ["Math", "Physics", "Programming"]
empty_list = []
empty_list.append(30)
empty_list.append(20)
print(empty_list)
````

### Tuples

**Definition**: A tuple is an ordered, immutable collection of elements.

**Characteristics:**
* Preserves order.
* Allows duplicates.
* Elements cannot be changed after creation (IMMUTABLE).
* Indexed.
* Use Case: When you want to ensure that data cannot be changed (e.g., coordinates, fixed records).

````python
my_tuple = (1, 2, 3, "A")
my_tuple
````

````python
print(my_tuple.index(1)) # return the index position of the value 
print(my_tuple.index(3)) #index in tuples are [0,1,2]
````

###  Sets
What is a set?
* A set is an unordered collection of UNIQUE elements.
* It cannot have duplicates.
* Elements can be of different data types (numbers, strings, etc.).
* Sets are mutable, so you can add or remove elements, but the elements themselves must be hashable (so no lists or dictionaries inside a set).

````python
my_set = {1, "apple", "Islam"}
# my_set.add("love")
my_set.remove('apple')
print(my_set)
````

### While Loop with List

````python
my_list = [1, 5, 3, 4]
i = 0
while i < len(my_list):
    print(my_list[i])
    i += 1 # same as i=i+1
````

### For Loop

Range function:
* start → where to begin (inclusive)
* stop → where to stop (exclusive)
* step → how much to increase each time

````python
# Range function --> range(start, stop, step)
for i in range(0, 10, 2):
    print(i)
````

### Looping Over List

````python
my_list = [1, 5, 4]
for element in my_list:
    print(element)
````

###  Continue Statement

````python
for i in range(10):
    if i == 5:
        continue # just skips if finds the value in the if statemetns
    print(i)
````

### Break Statement

````python
for i in range(11, 1000):
    if i % 11 == 0:
        print(i)
        break
````

### Challenge 3
Write a python script that calculates the factorial of a given number.

````python
def factorial(n):
    result = 1
    for i in range(2, n+1):
        result *= i
    return result

print(factorial(5))  # Output: 120
````

## In-class Activity: Challenge

A micro-lending start-up checks the day’s loan-repayment amounts for outliers caused by typos or fraud.

You junior developers must be able to:
Trace how Python data structures (lists, slicing, comprehensions) are used, and
Reason about a very small statistical sample that may hide or reveal anomalous values.


````python
import random
from statistics import mean, stdev

def flag_anomalies(amounts: list[int], student_id: str) -> list[int]:
    """
    Flags repayments whose distance from μ exceeds 1.5·σ of a 10 % random sample.
    A student-specific seed (last 3 ID digits) personalises the result.
    """
    # 1️⃣ Personalised seed -------------------------------------------------
    random.seed(int(student_id[-3:]))

    # 2️⃣ Tiny sample (10 % of the data; never less than 1 item) ------------
    sample = random.sample(amounts, k=max(1, len(amounts) // 10))

    # 3️⃣ Statistical threshold --------------------------------------------
    μ = mean(sample)
    σ = stdev(sample) if len(sample) > 1 else 0          # σ=0 if sample size = 1

    # 4️⃣ Flag anomalies -----------------------------------------------------
    return [a for a in amounts if abs(a - μ) > 1.5 * σ]

# --------------------------------------------------------------------------
transactions = [102,  98,  97,  99, 101, 100, 250,
                 97,  95, 420, 101,  99]

result = flag_anomalies(transactions, student_id="872 319")   # ← replace with YOUR ID
print(result)
````
